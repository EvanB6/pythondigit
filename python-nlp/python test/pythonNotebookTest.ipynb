```python
combined_text_season1 = ""
combined_text_season2 = ""


collPath = './group1'
for file in os.listdir(collPath):
    if file.endswith(".txt"):
        filepath = f"{collPath}/{file}"
        name, extension = os.path.splitext(file)
        #print(name)
        with open(filepath, 'r', encoding='utf8') as f:
            combined_text_season1 += f.read() + " "
            
tokens = word_tokenize(combined_text_season1.lower())
POStagged = pos_tag(tokens)
target_POS = ["VB", "VBD", "VBG", "VBN", "VBP", "VBZ"]

verbs = [word for word, tag in POStagged if tag in target_POS]

for w in sorted(set(verbs)):
    lemma = wn.morphy(w, wn.VERB) 
    lemma = lemma if lemma else w 
    synsets = wn.synsets(lemma, pos=wn.VERB)
    pos = {synset.pos() for synset in synsets}
    if synsets:
        print(f"Word: {w} | Lemma: {lemma}")
        print(f"  POS: {pos} | Synsets (ambiguity): {len(synsets)}")


        
print("\n-------------------------------------")
print("SEASON 2 - AMBIGUOUS VERBS")
print("==============================\n")


collPath = './group2'
for file in os.listdir(collPath):
    if file.endswith(".txt"):
        filepath = f"{collPath}/{file}"
        name, extension = os.path.splitext(file)
        #print(name)
        with open(filepath, 'r', encoding='utf8') as f:
            combined_text_season2 += f.read() + " "

tokens2 = word_tokenize(combined_text_season2.lower())
POStagged2 = pos_tag(tokens2)

verbs2 = [word for word, tag in POStagged2 if tag in target_POS]



for w in sorted(set(verbs2)):
    lemma = wn.morphy(w, wn.VERB)
    lemma = lemma if lemma else w
    
    synsets = wn.synsets(lemma, pos=wn.VERB)
    pos = {s.pos() for s in synsets}

    if synsets:
        print(f"Word: {w} | Lemma: {lemma}")
        print(f"  POS: {pos} | Synsets (ambiguity): {len(synsets)}")



from nltk.corpus import wordnet as wn

def get_ambiguity(word):
    lemma = wn.morphy(word, wn.VERB)
    lemma = lemma if lemma else word
    return len(wn.synsets(lemma, pos=wn.VERB))

def get_score(item):
    return item[1]

#season2 top 15

season1_scores = []

for w in set(verbs):
    score = get_ambiguity(w)
    season1_scores.append((w, score))

sorted_season1 = sorted(season1_scores, key=get_score, reverse=True)
season1_top15 = sorted_season1[:15]


#season2 top 15

season2_scores = []

for w in set(verbs2):
    score = get_ambiguity(w)
    season2_scores.append((w, score))

sorted_season2 = sorted(season2_scores, key=get_score, reverse=True)
season1_top15 = sorted_season2[:15]


output_file = "top_15_ambiguity_results.txt"

with open(output_file, "w", encoding="utf-8") as f:

    f.write("----------------------------\n")
    f.write("TOP 15 MOST AMBIGUOUS VERBS - SEASON 1\n")
    f.write("----------------------------\n\n")

    for word, score in season1_top15:
        f.write(f"{word:15} | Synsets: {score}\n")

    f.write("\n\n----------------------------\n")
    f.write("TOP 15 MOST AMBIGUOUS VERBS - SEASON 2\n")
    f.write("----------------------------\n\n")

    for word, score in season2_top15:
        f.write(f"{word:15} | Synsets: {score}\n")

print("Output written to:", output_file)
        
```
